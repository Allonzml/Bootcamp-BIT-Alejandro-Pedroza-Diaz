# 1. Registro de estudiantes

```python
print("Creación de Registro de Estudiantes")
estudiantes = [("Ana", 20, 4.5),("Carlos",22, 1.8),("Sofia",19, 4.9),
("Mateo",21, 2.8),("Lucía",20, 4.2),("David",23, 3.5),("Valentina",20, 4.6),("Andrés",22, 2.9),("Ricardo", 25,4.3),("Victoria",18,5.0),("Bernardo",27,4.5),("Luis Maria",24,4.8)]
```

```python
print(f"Total de estudiantes: {len(estudiantes)}")
```

```python
estudiantes
```

#  2. Mostrar la información

```python
print("Información General ")
print(" ")
for estudiante in estudiantes:
  Nombre, Edad, Nota = estudiante
  print(f"{Nombre} tiene {Edad} años y una nota de {Nota}")
  print(" "*5)
```

# 3. Clasificación de estudiantes

```python
print("Clasificación de Estudiantes según nota")
print(" ")

for Nombre, Edad, Nota in estudiantes:
  if Nota >=4.5:
    clasificación="Excelente"
  elif Nota >=4.0 and Nota <=4.49:
    clasificación="Bueno"
  elif Nota >=3.0 and Nota <=3.99:
    clasificación="Aceptable"
  else:
    clasificación="Reprobó"

  print(f"Estudiante {Nombre} tiene una clasificación de :{clasificación} a causa de su Nota: {Nota}")
  print(" "*5)
```

# 4. Promedio general

```python
suma_notas= 0
for _, _, Nota in estudiantes:
  suma_notas = suma_notas + Nota

promedio_general = suma_notas / len(estudiantes)

print(f"El promedio general de notas de los estudiantes es: {promedio_general}")
```

# 5. Búsqueda de estudiante

```python
print(" Mapeo de estudiante ")
print(" ")
buscar_nombre = input("Ingresa el nombre del estudiante que deseas buscar: ")

encontrado = False
for nombre, edad, nota in estudiantes:
    if nombre.lower() == buscar_nombre.lower():
        print(f"¡El estudiante '{nombre}' fue encontrado en la lista! (Edad: {edad}, Nota: {nota})")
        encontrado = True
        break

if not encontrado:
    print(f"No se encontró ningún estudiante con el nombre '{buscar_nombre}'.")
print("-" * 50)

```

# 6. Diccionario de ciudades

```python
print("Creación de Registro de las Ciudades de los Estudiantes")
ciudades_estudiantes = {
    "Ana": "Bogotá",
    "Carlos": "Medellín",
    "Sofia": "Bogotá",
    "Mateo": "Cali",
    "Lucía": "Pereira",
    "David": "Medellín",
    "Valentina": "Bogotá",
    "Andrés": "Armenia",
    "Ricardo": "Pereira",
    "Victoria": "Bogotá",
    "Bernardo": "Medellín",
    "Luis Maria": "Pasto"
}
```

```python
ciudades_estudiantes
```

```python
for estudiante, ciudad in ciudades_estudiantes.items():
    print(f"{estudiante} vive en : {ciudad}")
    print("-" *1)
```

# 7. Cantidad de estudiantes por ciudad

```python

conteo_ciudades = {}
for ciudad in ciudades_estudiantes.values():
    if ciudad in conteo_ciudades:
        conteo_ciudades[ciudad] += 1
    else:
        conteo_ciudades[ciudad] = 1
print("Cantidad de estudiaes por ciudad")
print(" ")
print(conteo_ciudades)
```

# 8. Ciclo While

```python

print("Ingresa números (el programa finaliza únicamente al ingresar 0):")

cantidad_numeros = 0
suma_total = 0

while True:
    numero_ingresado = float(input("Ingresa un número: "))

    if numero_ingresado == 0:
        break

    suma_total += numero_ingresado
    cantidad_numeros += 1

if cantidad_numeros > 0:
    promedio_numeros = suma_total / cantidad_numeros
else:
    promedio_numeros = 0

print("\nResultados del ciclo while:")
print(f"• Cantidad de números ingresados (sin contar el 0): {cantidad_numeros}")
print(f"• Suma total: {suma_total}")
print(f"• Promedio de los números ingresados: {promedio_numeros:.2f}")
print("-" * 50)
```

# 9. Uso de break y continue

```python
for numero in range(1, 31):
    if numero % 3 == 0:
        continue

    if numero == 25:
        break

    print(f"Número procesado: {numero}")
```
