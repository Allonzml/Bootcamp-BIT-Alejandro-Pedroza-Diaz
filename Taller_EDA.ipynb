**5.3• Menciona patrones, tendencias, relaciones o comportamientos importantes identificados en el conjunto de datos.**


    1.Tendencia de saturación o sobreuso del tiempo digital.
  
  La saturación del tiempo digital entre los usuarios activos en redes sociales, revela una notable homogeneidad de consumo: no existe una red social "marginal" o de uso secundario, ya que todas las plataformas principales (Snapchat, TikTok, Reddit, Instagram, YouTube, LinkedIn y X) registran un promedio de uso elevado que oscila en un margen muy estrecho de 5.33 a 5.58 horas diarias. Esto evidencia que el uso intensivo de redes sociales constituye un fenómeno transversal y sistémico en la población estudiantil, donde la elección de la plataforma específica no altera significativamente la carga horaria total dedicada al entorno digital.

    2.Patrón de deterioro simultáneo: Sueño y Salud Mental por consumo de Redes Sociales.
La relación entre el consumo de pantalla a altas horas de la noche, el recorte de las horas de descanso y la caída en los índices de bienestar resulta evidente en los datos: los estudiantes con consumo nocturno (Late_Night_Usage) duermen en promedio 6.24 horas, frente a las 7.31 horas de quienes no usan redes de noche, concentrando su densidad de descanso en un rango crítico de 5.5 a 6.5 horas. Este mismo grupo registra además una caída notable en su índice de salud mental, pasando de 84.21 a 75.88 puntos. Estos hallazgos evidencian que el uso nocturno actúa como un interruptor de los ciclos circadianos y un estimulante cognitivo que compromete la higiene del sueño.

    3.Comportamiento e incremento del estrés por horas de uso
La tendencia del estrés percibido (Perceived_Stress_Score) revela que este no escala de forma leve, sino que presenta un salto drástico al pasar de un consumo moderado a uno intensivo: en el uso bajo (2-3 horas), la mediana de estrés es de 8.0 puntos; en el uso moderado (3-6 horas), sube a 12.0 puntos (+50%); y en el uso alto (6 horas), se dispara a 20.0 puntos (+150% respecto al uso bajo). Este patrón indica que superar las 6 horas de uso diario marca un punto de inflexión donde la sobrecarga de información y la interacción digital constante se traducen en niveles elevados de tensión psicológica.

    6.Conclusión General

El estudio demuestra que el impacto de las redes sociales en los estudiantes no depende de la plataforma elegida ni del nivel educativo, sino de la cantidad de horas y el momento del día en que se consumen.

El uso intensivo (más de 6 horas diarias) y el consumo nocturno desplazan directamente el tiempo de estudio y el descanso, reduciendo el sueño a un promedio crítico de 6.2 horas, duplicando el estrés percibido y erosionando el rendimiento académico. En consecuencia, la desconexión nocturna y la autorregulación horaria resultan fundamentales para preservar la salud mental y el desempeño escolar de los estudiantes.
